## Introduction and Strategic Overview
The global population is rapidly aging, driven by increasing life expectancies and declining fertility rates. According to the World Health Organization (WHO, 2026). Aged 60 and over reached 1.1 billion people in 2023 and is projected to nearly double to 2.1 billion by 2050.

As the older population grows, addressing mental health conditions, particularly depression, becomes increasingly important. However, Mental health conditions in the elderly population often go unrecognized especially in relation to depressive symptoms which often goes untreated creating challenges for healthcare organisations.

Healthcare organisations face challenges in identifying risk factors such as social isolation, chronic illnesses and financial difficulties which are often related to depressive symptoms.

This project will examine demographics, socioeconomic and health related risk factors that are associated with the depressive symptoms. Using the NHANES August 2021–August 2023 dataset(CDC, 2024; NHANES, 2026) and PHQ-9 responses(Kroenke, Spitzer and Williams, 2001). By applying predictive analytics to explore and identify high risk groups, support screening priorities and planning resource allocation. The goal of the project is not to predict future onset or provide clinical diagnostics.
